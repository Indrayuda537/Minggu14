```python
# ============================================================
# KLASIFIKASI CITRA DENGAN CNN DAN TRANSFER LEARNING
# FULL PROJECT SESUAI INSTRUKSI TUGAS
# CIFAR-10 + CNN + TRANSFER LEARNING + GRADCAM + TSNE + ROC
# ============================================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import time

from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.applications import (
    MobileNetV2,
    VGG16,
    ResNet50
)

from tensorflow.keras.optimizers import Adam, SGD

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_curve,
    auc
)

from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE

# ============================================================
# 1. LOAD DATASET CIFAR-10
# ============================================================

print("\nLOADING CIFAR-10 DATASET...")

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# normalisasi
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

num_classes = 10

# untuk ROC multiclass
y_test_bin = label_binarize(y_test, classes=range(num_classes))

print("Train Shape :", x_train.shape)
print("Test Shape  :", x_test.shape)

# ============================================================
# 2. DATA AUGMENTATION
# ============================================================

print("\nMEMBUAT DATA AUGMENTATION...")

datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    zoom_range=0.2,
    shear_range=0.2,
    fill_mode='nearest'
)

datagen.fit(x_train)

# ============================================================
# VISUALISASI AUGMENTASI
# ============================================================

sample = x_train[0].reshape((1,32,32,3))

plt.figure(figsize=(8,8))

i = 0

for batch in datagen.flow(sample, batch_size=1):

    plt.subplot(3,3,i+1)
    plt.imshow(batch[0])
    plt.axis("off")

    i += 1

    if i == 9:
        break

plt.suptitle("Visualisasi Data Augmentation")
plt.show()

# ============================================================
# 3. CNN FROM SCRATCH
# ============================================================

def build_cnn(optimizer):

    model = models.Sequential([

        layers.Input(shape=(32,32,3)),

        layers.Conv2D(32, (3,3), activation='relu'),
        layers.MaxPooling2D(2,2),

        layers.Conv2D(64, (3,3), activation='relu'),
        layers.MaxPooling2D(2,2),

        layers.Conv2D(128, (3,3), activation='relu'),

        layers.Flatten(),

        layers.Dense(256, activation='relu'),

        layers.Dropout(0.5),

        layers.Dense(num_classes, activation='softmax')
    ])

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# ============================================================
# EKSPERIMEN OPTIMIZER
# ============================================================

adam_optimizer = Adam(learning_rate=0.001)
sgd_optimizer = SGD(learning_rate=0.01)

# ============================================================
# TRAIN CNN ADAM
# ============================================================

print("\nTRAINING CNN (ADAM)...")

start_time = time.time()

cnn_model = build_cnn(adam_optimizer)

history_cnn = cnn_model.fit(
    datagen.flow(x_train, y_train, batch_size=64),
    epochs=10,
    validation_data=(x_test, y_test),
    verbose=1
)

cnn_training_time = time.time() - start_time

# ============================================================
# TRAIN CNN SGD
# ============================================================

print("\nTRAINING CNN (SGD)...")

cnn_sgd = build_cnn(sgd_optimizer)

history_sgd = cnn_sgd.fit(
    datagen.flow(x_train, y_train, batch_size=64),
    epochs=5,
    validation_data=(x_test, y_test),
    verbose=1
)

# ============================================================
# 4. TRANSFER LEARNING
# ============================================================

def build_transfer_model(base_model):

    base_model.trainable = False

    x = layers.GlobalAveragePooling2D()(base_model.output)

    x = layers.Dense(128, activation='relu')(x)

    x = layers.Dropout(0.5)(x)

    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = models.Model(base_model.input, outputs)

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# ============================================================
# MOBILENETV2
# ============================================================

print("\nTRAINING MOBILENETV2...")

mobilenet_base = MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(32,32,3)
)

mobilenet = build_transfer_model(mobilenet_base)

history_mobilenet = mobilenet.fit(
    datagen.flow(x_train, y_train, batch_size=64),
    epochs=5,
    validation_data=(x_test, y_test),
    verbose=1
)

# ============================================================
# VGG16
# ============================================================

print("\nTRAINING VGG16...")

vgg_base = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(32,32,3)
)

vgg = build_transfer_model(vgg_base)

history_vgg = vgg.fit(
    datagen.flow(x_train, y_train, batch_size=32),
    epochs=5,
    validation_data=(x_test, y_test),
    verbose=1
)

# ============================================================
# RESNET50
# ============================================================

print("\nTRAINING RESNET50...")

resnet_base = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(32,32,3)
)

resnet = build_transfer_model(resnet_base)

history_resnet = resnet.fit(
    datagen.flow(x_train, y_train, batch_size=32),
    epochs=5,
    validation_data=(x_test, y_test),
    verbose=1
)

# ============================================================
# 5. FINE TUNING MOBILENET
# ============================================================

print("\nFINE TUNING MOBILENETV2...")

mobilenet.trainable = True

for layer in mobilenet.layers[:-20]:
    layer.trainable = False

mobilenet.compile(
    optimizer=Adam(1e-5),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_finetune = mobilenet.fit(
    datagen.flow(x_train, y_train, batch_size=64),
    epochs=5,
    validation_data=(x_test, y_test),
    verbose=1
)

# ============================================================
# 6. EVALUASI MODEL
# ============================================================

def evaluate_model(model, name):

    print("\n===================================")
    print("EVALUASI :", name)
    print("===================================")

    start_inf = time.time()

    predictions = model.predict(x_test)

    inference_time = time.time() - start_inf

    y_pred = np.argmax(predictions, axis=1)

    # confusion matrix
    print("\nCONFUSION MATRIX")
    print(confusion_matrix(y_test, y_pred))

    # classification report
    print("\nCLASSIFICATION REPORT")
    print(classification_report(y_test, y_pred))

    # ROC AUC
    roc_auc = auc(
        *roc_curve(
            y_test_bin.ravel(),
            predictions.ravel()
        )[:2]
    )

    print("\nROC-AUC :", roc_auc)

    print("Inference Time :", inference_time, "seconds")

    return predictions

cnn_predictions = evaluate_model(cnn_model, "CNN SCRATCH")
mobilenet_predictions = evaluate_model(mobilenet, "MOBILENETV2")
vgg_predictions = evaluate_model(vgg, "VGG16")
resnet_predictions = evaluate_model(resnet, "RESNET50")

# ============================================================
# 7. LEARNING CURVE
# ============================================================

def plot_learning_curve(history, title):

    plt.figure(figsize=(12,5))

    # accuracy
    plt.subplot(1,2,1)

    plt.plot(history.history['accuracy'], label='Train')
    plt.plot(history.history['val_accuracy'], label='Validation')

    plt.title(title + " Accuracy")
    plt.legend()

    # loss
    plt.subplot(1,2,2)

    plt.plot(history.history['loss'], label='Train')
    plt.plot(history.history['val_loss'], label='Validation')

    plt.title(title + " Loss")
    plt.legend()

    plt.show()

plot_learning_curve(history_cnn, "CNN Scratch")
plot_learning_curve(history_mobilenet, "MobileNetV2")
plot_learning_curve(history_vgg, "VGG16")
plot_learning_curve(history_resnet, "ResNet50")

# ============================================================
# 8. FEATURE MAP VISUALIZATION
# ============================================================

print("\nVISUALIZING FEATURE MAPS...")

feature_model = tf.keras.Model(
    inputs=cnn_model.inputs,
    outputs=cnn_model.layers[1].output
)

feature_maps = feature_model.predict(x_test[:1])

plt.figure(figsize=(12,12))

for i in range(16):

    plt.subplot(4,4,i+1)

    plt.imshow(feature_maps[0,:,:,i], cmap='viridis')

    plt.axis('off')

plt.suptitle("Feature Maps")
plt.show()

# ============================================================
# 9. FILTER VISUALIZATION
# ============================================================

print("\nVISUALIZING FILTERS...")

# cari layer Conv2D pertama
conv_layer = None

for layer in cnn_model.layers:

    if isinstance(layer, tf.keras.layers.Conv2D):

        conv_layer = layer
        break

# ambil filter
filters, biases = conv_layer.get_weights()

print("Filter Shape :", filters.shape)

plt.figure(figsize=(10,10))

# tampilkan maksimal 16 filter
num_filters = min(16, filters.shape[-1])

for i in range(num_filters):

    f = filters[:, :, :, i]

    plt.subplot(4,4,i+1)

    # tampilkan channel pertama
    plt.imshow(f[:, :, 0], cmap='gray')

    plt.axis('off')

plt.suptitle("Filter Visualization")

plt.show()

# ============================================================
# 10. GRAD-CAM
# ============================================================

print("\nGENERATING GRAD-CAM...")

def gradcam(model, img):

    # pastikan model sudah build
    model.predict(img)

    # cari conv layer terakhir
    last_conv_layer = None

    for layer in reversed(model.layers):

        if isinstance(layer, tf.keras.layers.Conv2D):

            last_conv_layer = layer.name
            break

    # grad model
    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[
            model.get_layer(last_conv_layer).output,
            model.outputs[0]
        ]
    )

    with tf.GradientTape() as tape:

        conv_outputs, predictions = grad_model(img)

        pred_index = tf.argmax(predictions[0])

        loss = predictions[:, pred_index]

    grads = tape.gradient(loss, conv_outputs)

    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0,1,2)
    )

    conv_outputs = conv_outputs[0]

    heatmap = tf.reduce_sum(
        pooled_grads * conv_outputs,
        axis=-1
    )

    heatmap = tf.maximum(heatmap, 0)

    max_heat = tf.reduce_max(heatmap)

    if max_heat > 0:
        heatmap /= max_heat

    return heatmap.numpy()

sample_img = x_test[:1]

heatmap = gradcam(cnn_model, sample_img)

plt.figure(figsize=(6,6))

plt.imshow(heatmap, cmap='jet')

plt.title("Grad-CAM Heatmap")

plt.colorbar()

plt.show()

# ============================================================
# 11. t-SNE VISUALIZATION
# ============================================================

print("\nVISUALIZING t-SNE...")

embedding_model = tf.keras.Model(
    inputs=cnn_model.inputs,
    outputs=cnn_model.layers[-3].output
)

features = embedding_model.predict(x_test[:1000])

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

tsne_result = tsne.fit_transform(features)

plt.figure(figsize=(10,8))

scatter = plt.scatter(
    tsne_result[:,0],
    tsne_result[:,1],
    c=y_test[:1000].flatten(),
    cmap='tab10'
)

plt.title("t-SNE Feature Embedding")

plt.colorbar(scatter)

plt.show()

# ============================================================
# 12. ROC CURVE
# ============================================================

def plot_roc_curve(model, name):

    predictions = model.predict(x_test)

    plt.figure(figsize=(8,6))

    for i in range(num_classes):

        fpr, tpr, _ = roc_curve(
            y_test_bin[:,i],
            predictions[:,i]
        )

        roc_auc = auc(fpr, tpr)

        plt.plot(
            fpr,
            tpr,
            label=f"Class {i} AUC={roc_auc:.2f}"
        )

    plt.plot([0,1],[0,1],'k--')

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title("ROC Curve - " + name)

    plt.legend()

    plt.show()

plot_roc_curve(cnn_model, "CNN Scratch")
plot_roc_curve(mobilenet, "MobileNetV2")
plot_roc_curve(vgg, "VGG16")
plot_roc_curve(resnet, "ResNet50")

# ============================================================
# 13. TRAINING TIME
# ============================================================

print("\nCNN Training Time :", cnn_training_time, "seconds")

print("\nPROGRAM SELESAI")
```
